```python
import os
import glob
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. البحث التلقائي عن أي ملف PDF مرفوع في كولاب
pdf_files = glob.glob("*.pdf")

if not pdf_files:
    print("❌ لم يتم العثور على أي ملف PDF في كولاب!")
    print("📂 الملفات الموجودة حالياً في المجلد هي:", os.listdir('.'))
    raise FileNotFoundError("⚠️ تأكدي من رفع ملف الـ PDF من القائمة الجانبية اليسرى (أيقونة المجلد 📁) في Colab ثم أعيدي تشغيل الخلية.")

# اختيار أول ملف PDF يتم العثور عليه تلقائياً
pdf_path = pdf_files[0]
print(f"📄 تم العثور على ملف الـ PDF: {pdf_path}")

# 2. تحميل المستند
loader = PyPDFLoader(pdf_path)
documents = loader.load()
print(f"✅ تم تحميل المستند بنجاح! إجمالي عدد الصفحات: {len(documents)}")

# 3. تقسيم النص إلى قطع (Chunks)
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,      # حجم القطعة النصية
    chunk_overlap=200     # التداخل بين القطع لعدم فقدان المعنى والسياق
)

chunks = text_splitter.split_documents(documents)
print(f"✅ تم تقسيم المستند بنجاح إلى {len(chunks)} قطعة نصية (Chunk).")
```

```python
# 1. تثبيت مكتبة sentence-transformers للـ Embeddings
!pip install -q sentence-transformers

from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# 2. تحميل نموذج الـ Embeddings السريع والفعال
print("⏳ جاري تحميل نموذج الـ Embeddings...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# 3. تحويل الـ Chunks إلى متجهات وتخزينها داخل FAISS
print("⏳ جاري تحويل النصوص إلى متجهات (Embeddings) وبناء قاعدة البيانات FAISS...")
vectorstore = FAISS.from_documents(chunks, embeddings)

print("✅ تم إنشاء الـ Vector Database (FAISS) بنجاح وتخزين البيانات!")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# 1. تهيئة الـ Retriever لجلب أفضل 3 مقاطع ذات صلة
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

# 2. تهيئة نموذج Gemini
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0.2
)

# 3. إعداد الـ Prompt المخصص للـ Chatbot
prompt_template = """You are an AI assistant designed to answer questions about the DStarix Internship Guide.
Use the following retrieved context to answer the user's question accurately.
If the answer is not contained in the context, state clearly that the information is not available in the document.

Context:
{context}

Question:
{question}
"""

prompt = ChatPromptTemplate.from_template(prompt_template)

# دالة لتنسيق النصوص المسترجعة
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# 4. بناء الـ RAG Chain بالأسلوب الحديث (LCEL)
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

print("✅ تم بناء الـ RAG Chain بنجاح وتجهيز الـ Chatbot!")
```

```python
print("🤖 أهلاً بك في DStarix Internship Guide Chatbot!")
print("اكتبي سؤالك واضغطي Enter (اكتبي 'exit' للإنهاء):\n")

while True:
    user_query = input("أنتِ: ")
    if user_query.lower() in ['exit', 'quit', 'إنهاء', 'خروج']:
        print("🤖 Chatbot: بالتوفيق في التدريب! باي باي 👋")
        break

    if not user_query.strip():
        continue

    print("\n⏳ جاري البحث والتفكير...")
    response = rag_chain.invoke(user_query)
    print(f"\n🤖 Chatbot:\n{response}\n")
    print("-" * 50)
```
